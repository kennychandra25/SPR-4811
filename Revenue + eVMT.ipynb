```python
chtime_data= {0: [], 1: [], 2: [], 3: [], 4: []}
# Grouping based on same station
for t,scenario in enumerate(ch_data):
    for data_year in scenario:
        # Grouping by chargerID and chargerType, summing the chargingTime_min
        data = pd.DataFrame(data_year.groupby(['chargerID', 'chargerType'], sort=True)['chargingTime_min'].sum())
        data = data.reset_index()

        data = data.set_index('chargerID').join(data_year.set_index('chargerID')[['EV Level2', 'EV DC Fast']], how='left')
        data = data.reset_index()
        chtime_data[t].append(data)

    print(chtime_data[t][-1].head())
```

```python
# Define a function to calculate energy based on chargerType
def calculate_energy(row):
    if row['chargerType'] == 'L2':
        return row['chargingTime_min_yr']/60 * 12
    elif row['chargerType'] == 'L3':
        return row['chargingTime_min_yr']/60 * 50
    elif row['chargerType'] == 'L2 and L3':
        return row['chargingTime_min_yr']/60 * (12*row['EV Level2']/(row['EV Level2'] + row['EV DC Fast']) + 50*row['EV DC Fast']/(row['EV Level2'] + row['EV DC Fast']))
    else:
        return 0  # or handle other cases as needed

for scenario, data_year in chtime_data.items():
    for data in data_year:
        data['energy_kwh_yr'] = data.apply(calculate_energy, axis=1)
```

```python
# Define a function to calculate energy based on chargerType
def calculate_revenue(row):
    if row['chargerType'] == 'L2':
        return row['energy_kwh_yr'] * 0.25
    elif row['chargerType'] == 'L3':
        return row['energy_kwh_yr'] * 0.35
    elif row['chargerType'] == 'L2 and L3':
        return row['energy_kwh_yr'] * (0.25*row['EV Level2']/(row['EV Level2'] + row['EV DC Fast']) + 0.35*row['EV DC Fast']/(row['EV Level2'] + row['EV DC Fast']))
    else:
        return 0  # or handle other cases as needed

for scenario, data_year in chtime_data.items():
    for data in data_year:
        data['revenue_$'] = data.apply(calculate_revenue, axis=1)
```

```python
plt.rcParams["font.family"] = "times"
sns.set_style("ticks")

titles = ["Year 2024", "Year 2025", "Year 2026", "Year 2028", "Year 2031"]

# Map for legend labels
scenario_legend_labels = {'Scenario A': 'A', 'Scenario B': 'B', 'Scenario C': 'C', 'Scenario D': 'D', 'Scenario E': 'E'}

fig, axes = plt.subplots(1, 5, figsize=(14, 4.5), sharey=True, sharex=True)

for i in range(5):
    plot_data_list = []
    for scenario, data_year in chtime_data.items():
        datatime = data_year[i]
        temp_data = datatime[['revenue_$', 'chargerType']].copy()
        temp_data['Scenario'] = scenario_labels[scenario]
        plot_data_list.append(temp_data)
    plot_data = pd.concat(plot_data_list, ignore_index=True)
    plot_data.rename(columns={'revenue_$': 'Revenue ($)', 'chargerType': 'Charger Type'}, inplace=True)
    plot_data = plot_data[plot_data['Charger Type'].isin(['L2', 'L3', 'L2 and L3'])]

    sns.boxplot(
        x='Charger Type', y='Revenue ($)', hue='Scenario', data=plot_data, ax=axes[i],
        linewidth=0.7, linecolor='k', showfliers=False,
        palette='Set2', width=0.7, saturation=0.85, showmeans=True,
        meanprops={'marker':'o','markerfacecolor':'k','markeredgecolor':'k','markersize':'3'}
    )
    axes[i].set_title(f'{titles[i]}', fontweight="bold", fontsize=16)
    axes[i].set_xlabel('Charger Type' if i == 2 else '', fontweight="bold", fontsize=16, labelpad=9)
    axes[i].set_ylabel('Revenue ($)', fontweight="bold", fontsize=16 if i == 0 else 0, labelpad=9)
    axes[i].grid(axis='y', linestyle='--', alpha=0.7)
    axes[i].tick_params(axis="both", color="k", labelsize=14)
    axes[i].set_facecolor("whitesmoke")
    if i != 0:
        axes[i].legend_.remove()

# Change legend labels to A, B, C, D, E
legend = axes[0].legend(title="Scenario", title_fontsize=12, fontsize=12, loc="upper left", frameon=True, ncols=1)
for t, label in zip(legend.texts, [scenario_legend_labels.get(txt.get_text(), txt.get_text()) for txt in legend.texts]):
    t.set_text(label)
plt.setp(legend.get_title(), fontweight='bold')
legend.get_frame().set_edgecolor('black')
legend.get_frame().set_linewidth(1.0)
legend.get_frame().set_facecolor('white')
plt.tight_layout()
plt.savefig("Revenue_by_Charger_Type_Year.png", dpi=600, bbox_inches="tight")
plt.show()
```

```python
link_type_vmt = {0: [], 1: [], 2: [], 3: [], 4: []}
# Grouping total VMT by type for each scenario
for t, scenario in enumerate(metsr_link):
    for link in scenario:
        # Grouping by type and summing the total_vmt
        type_vmt = pd.DataFrame(link.groupby(['type'], sort=True)['total_vmt'].sum())
        type_vmt = type_vmt.reset_index()
        type_vmt = type_vmt.sort_values(by=['total_vmt'], ascending=False)
        link_type_vmt[t].append(type_vmt)
```
